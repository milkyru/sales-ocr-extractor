```python
#!/usr/bin/env python3
"""
Simple test script for the sales extractor
"""

import os
import sys

# Add the current directory to Python path so we can import our modules
sys.path.insert(0, os.path.dirname(os.path.abspath(__file__)))

def test_single_image(image_path):
    """Test extraction on a single image"""
    print(f"Testing sales extraction on: {image_path}")
    print("=" * 60)
    
    try:
        # Import our extractor
        from sales_report import SalesReportExtractor
        
        # Create extractor instance
        extractor = SalesReportExtractor()
        
        # Process the image
        print("🔄 Processing image...")
        data = extractor.process_image(image_path)
        
        print(f"✅ Successfully extracted {len(data)} transactions")
        
        # Display results
        if data:
            print("\n📊 EXTRACTED DATA:")
            print("-" * 40)
            for i, transaction in enumerate(data, 1):
                print(f"Transaction {i}:")
                for key, value in transaction.items():
                    print(f"  {key}: {value}")
                print()
        else:
            print("⚠️  No transactions extracted")
            print("This might be normal if the image has no transaction data")
        
        return data
        
    except FileNotFoundError as e:
        print(f"❌ File error: {str(e)}")
        print("📋 Make sure the image file exists and path is correct")
        
    except ImportError as e:
        print(f"❌ Import error: {str(e)}")
        print("📋 Make sure all required packages are installed:")
        print("   pip install pytesseract opencv-python pandas numpy pillow")
        
    except Exception as e:
        print(f"❌ Unexpected error: {str(e)}")
        print("📋 Try running the debug script first to identify the issue")
        
    return None

def save_results_to_csv(data, output_path="test_extraction.csv"):
    """Save extracted data to CSV for inspection"""
    if not data:
        print("No data to save")
        return
    
    try:
        import pandas as pd
        df = pd.DataFrame(data)
        df.to_csv(output_path, index=False)
        print(f"💾 Results saved to: {output_path}")
        
        # Display summary
        print("\n📈 EXTRACTION SUMMARY:")
        print("-" * 30)
        print(f"Total rows: {len(df)}")
        print(f"Columns: {', '.join(df.columns)}")
        
        # Show data quality
        missing_data = df.isnull().sum()
        if missing_data.any():
            print("\n⚠️  Missing data detected:")
            for col, count in missing_data.items():
                if count > 0:
                    print(f"  {col}: {count} missing values")
        else:
            print("✅ No missing data detected")
            
    except Exception as e:
        print(f"❌ Error saving CSV: {str(e)}")

def main():
    """Main function"""
    print("🧪 SALES EXTRACTOR TEST TOOL")
    print("=" * 40)
    
    # Get image path
    if len(sys.argv) > 1:
        image_path = sys.argv[1]
    else:
        print("Enter the path to your sales report image:")
        image_path = input("Image path: ").strip().strip('"\'')
    
    if not image_path:
        print("❌ No image path provided!")
        return
    
    # Check if file exists
    if not os.path.exists(image_path):
        print(f"❌ File not found: {image_path}")
        print("💡 Try using the full/absolute path to the image")
        return
    
    # Test extraction
    data = test_single_image(image_path)
    
    # Save results if successful
    if data:
        save_to_csv = input("\nSave results to CSV? (y/n): ").lower().strip()
        if save_to_csv == 'y':
            save_results_to_csv(data)
    
    print("\n🏁 Test completed!")

if __name__ == "__main__":
    main()
```
